# <font color='steelblue'>Scrapping de letras de música utilizando BeautifulSoup</font>


Esse código implementa uma rotina que faz busca de uma letra de música na plataforma letras.mus.br e retorna a letra da música.

Como exemplo, está sendo feita uma busca a partir da música blackbird dos Beatles.

É importante ressaltar que:

* um dos argumentos de busca é o nome do artista e o outro o nome da música. Você deve digitar o nome exato para que a busca tenha resultado. O código, entretanto, vai tratar a capitalização e os caracteres em branco.
* Você pode fazer alterações no código, conforme achar mais conveniente
* Ao final, são propostos alguns exercícios envolvendo listas, dicionários e strings, que devem ser implementados reaproveitando o código deste notebbok

```python
import requests
from bs4 import BeautifulSoup

def get_lyrics(artist, song):
    url = f"https://www.vagalume.com.br/{artist}/{song}.html"

    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, "html.parser")
        lyrics_div = soup.find("div", {"id": "lyrics"})

        if lyrics_div:
            lyrics = " ".join(lyrics_div.stripped_strings)
            lyrics = lyrics.encode('latin-1', 'ignore').decode('utf-8')
            return lyrics

    return None

```

```python
artist = "the beatles"
song = "blackbird"

lyrics = get_lyrics(artist.lower().replace(" ", "-"), song.lower().replace(" ", "-"))

if lyrics:
    print(lyrics)
else:
    print("Letras não encontradas.")
```
